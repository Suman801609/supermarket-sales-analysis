# Supermarket Sales Analysis

**Project:** Supermarket Sales Analysis  
**Tools:** Python, Pandas, Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Load Dataset

```python
df = pd.read_csv("data/supermarket_sales_clean.csv")

print("Dataset loaded!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 2. Explore Dataset

```python
print(df.head())
print("\nShape:", df.shape)
print("\nColumns:", df.columns.tolist())
```

## 3. Data Cleaning

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

## 4. Sales Analysis

```python
sales = df["Sales"].sum()
profit = df["Profit"].sum()
quantity = df["Quantity"].sum()

print("===== SALES ANALYSIS =====")
print("Total Sales:", round(sales, 2))
print("Total Profit:", round(profit, 2))
print("Total Quantity:", quantity)
print("Average Sales:", round(df["Sales"].mean(), 2))
print("Average Profit:", round(df["Profit"].mean(), 2))
```

## 5. Category Analysis

```python
category = df.groupby("Category")[["Sales", "Profit"]].sum()

print("===== CATEGORY ANALYSIS =====")
print(category.round(2))

category["Sales"].plot(kind="bar", title="Sales by Category")
plt.ylabel("Sales")
plt.show()
```

## 6. Region Analysis

```python
region = df.groupby("Region")[["Sales", "Profit"]].sum()

print("===== REGION ANALYSIS =====")
print(region.round(2))

region["Sales"].plot(kind="bar", title="Sales by Region")
plt.ylabel("Sales")
plt.show()
```

## 7. Year-wise Analysis

```python
df["Year"] = df["Order Date"].dt.year
year = df.groupby("Year")[["Sales", "Profit"]].sum()

print("===== YEAR-WISE ANALYSIS =====")
print(year.round(2))

year["Sales"].plot(kind="line", marker="o", title="Year-wise Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

## 8. Month-wise Analysis

```python
df["Month"] = df["Order Date"].dt.month
month = df.groupby("Month")[["Sales", "Profit"]].sum()

print("===== MONTH-WISE ANALYSIS =====")
print(month.round(2))

month["Sales"].plot(kind="line", marker="o", title="Month-wise Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

## 9. Top 10 Products

```python
top = df.groupby("Product Name")[["Sales", "Profit"]].sum()
top = top.sort_values("Sales", ascending=False).head(10)

print("===== TOP 10 PRODUCTS =====")
print(top.round(2))

top["Sales"].sort_values().plot(
    kind="barh",
    title="Top 10 Products by Sales"
)
plt.xlabel("Sales")
plt.show()
```

## 10. Profit Margin

```python
category["Profit_Margin_%"] = (
    category["Profit"] / category["Sales"] * 100
)

print("===== PROFIT MARGIN =====")
print(category[["Sales", "Profit", "Profit_Margin_%"]].round(2))

category["Profit_Margin_%"].plot(
    kind="bar",
    title="Profit Margin by Category"
)
plt.ylabel("Profit Margin (%)")
plt.show()
```

## 11. Final Summary

```python
print("===== FINAL SUMMARY =====")
print("Total Sales:", round(sales, 2))
print("Total Profit:", round(profit, 2))
print("Highest Sales Category:", category["Sales"].idxmax())
print("Highest Profit Category:", category["Profit"].idxmax())
print("Highest Sales Region:", region["Sales"].idxmax())
print("Highest Sales Year:", year["Sales"].idxmax())
print("Top Product:", top["Sales"].idxmax())
```
